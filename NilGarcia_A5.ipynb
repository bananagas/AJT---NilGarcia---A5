```python
class Song():
    def __init__(self, titol, autor, durada):
        self.titol = titol
        self.autor = autor
        self.durada = durada

    def info(self):
        return f"{self.titol} - {self.autor} ({self.durada} segons)"
        
        

class Playlist():
    def __init__(self, song_list):
        self.song_list = song_list

    def afegir_canco(self, song):
        self.song_list.append(song)

    def durada_total(self):
        duradamax = 0
        for i in self.song_list:
            duradamax += i.durada
            
        print(duradamax)

    def mostrar_playlist(self):
        for i in self.song_list:
            print(i.info())
            


    def canco_mes_llarga(self):
        cancogran = 0
        guanyador = ""
        for i in self.song_list:
            if i.durada > cancogran:
                cancogran = i.durada
                guanyador = i
                
        print(guanyador.info())
```

```python

s1 = Song("Espreso", "Sabrina Carpeta", 200)
s2 = Song("Song 2", "Blur", 210)


p1 = Playlist([])


p1.afegir_canco(s1)
p1.afegir_canco(s2)

s3 = Song("If I Am With You", "Jjk", 140)

p1.afegir_canco(s3)

p1.mostrar_playlist()
p1.durada_total()

p1.canco_mes_llarga()
```

```python
class Personatge():
    def __init__(self, nom, vida, força, nivell):
        self.nom = nom
        self.vida = vida
        self.força = força
        self.nivell = nivell

    def atacar(self, enemic):
        if enemic.vida <= self.força:
            enemic.vida = 0
            print(f"{self.nom} ha derrotat aquest enemic!")
            
        else:
            enemic.vida -= self.força
            print(f"{self.nom} ha restat {self.força} punts de vida al enemic!")

    def pujar_nivell(self):
        self.nivell += 1
        self.força += 2
        self.vida += 5
        print(f"El personatge {self.nom} ha pujat de nivell!")

    def esta_viu(self):
        if self.vida <= 0:
            return False
        else:
            return True
        
    def __str__(self):
        return f"El personatge {self.nom} te {self.vida} punts de vida, {self.força} de força i és nivell {self.nivell}."

```

```python

h1 = Personatge("Cloud", 15, 3, 1)
print(h1)

h2 = Personatge("Moogle", 10, 2, 1)
print(h2)

while h1.esta_viu() and h2.esta_viu():
    h1.atacar(h2)
    h2.atacar(h1)


if h1.esta_viu():
    h1.pujar_nivell()
    print(h1)
else:
    h2.pujar_nivell()
    print(h2)

```
